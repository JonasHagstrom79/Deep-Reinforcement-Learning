```python
import gymnasium as gym
```

```python
env = gym.make("CartPole-v1", render_mode="rgb_array")
#env = gym.wrappers.RecordVideo(env, "video") # video recording
env = gym.wrappers.RecordEpisodeStatistics(env) # episode statistics
observation, info = env.reset()
print("obs:", observation)
print("info:", info)
G = 0

# terminated means the same thing as "done"
terminated = False # episode ends
truncated = False #
while not terminated and not truncated:
  action = env.action_space.sample()
  observation, reward, terminated, truncated, info = env.step(action)
  G += reward
  print("obs:", observation)
  print("reward:", reward)
  print("terminated:", terminated)
  print("truncated:", truncated)
  print("info:", info)


# close environment
env.close()
print("G:", G)
```

```python
def make_env():
  env = gym.make("CartPole-v1", render_mode="rgb_array")
  env = gym.wrappers.RecordEpisodeStatistics(env)
  return env
```

```python
num_envs = 16
envs = gym.vector.SyncVectorEnv(
    [make_env for _ in range(num_envs)])
```

```python
# try with vector env
observation, infos = envs.reset()
print("obs:", observation)
print("info:", infos)

# terminated means the same thing as "done"
terminated = False # episode ends
truncated = False #
while not terminated and not truncated:
  actions = envs.action_space.sample()
  next_obs, rewards, terminateds, truncateds, infos = envs.step(actions)
  print("next_obs:", next_obs)
  print("reward:", rewards)
  print("terminated:", terminateds)
  print("truncated:", truncateds)
  print("info:", infos)

  # let's stop whenever the first env episode ends
  terminated = terminateds[0]
  truncated = truncateds[0]

# close environment
env.close()
print("G:", G)
```
